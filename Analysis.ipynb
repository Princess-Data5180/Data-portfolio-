```python
df = df.drop_duplicates()
```

```python
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False).head(10)
```

```python
df= df.drop(columns=["Location Point1"])
```

```python
df = df.dropna(subset=["CAMIS","INSPECTION DATE", "DBA"])
```

```python
df.shape
```

# 1. TOP VIOLATIONS 

```python
df["VIOLATION DESCRIPTION"].value_counts().head(10)
```

```python
df["VIOLATION DESCRIPTION"].value_counts().head(10).plot(kind="barh",figsize=(8,5), title="TOP 10 VIOLATIONS")
```

#### INSIGHT:
The most common violations relate to food safety and hygiene practices,indicating recurring compliance issues across NYC restaurants.

# 2. TOP RESTAURANTS

```python
df["DBA"].value_counts().head(10)
```

```python
df["DBA"].value_counts().head(10).plot(kind="barh", figsize=(8,5), title="Most Inspections")
```

#### INSIGHT:
A small number of restaurants appear frequently in inspection records, likely due to repeated inspections or higher risk classificatios.

# 3. SCORE SUMMARY

```python
df["SCORE"].describe()
```

```python
df["SCORE"].plot(kind="hist", bins=30,title="Score Distribution")
```

#### INSIGHT:
Most restaurant inspection scores fall within a moderate range, showing that while extreme violatons exists, the majority of restaurants are within acceptable compliance levels 

# 4. GRADE DISTRIBUTION

```python
df["GRADE"].value_counts()
```

```python
ax =df["GRADE"].value_counts().plot(kind="barh",figsize=(8,5))
ax.set_title("Grade Distribution Of Restaurants")
ax.set_xlabel("Number Of Restaurants")
ax.set_ylabel("Grade")
```

#### INSIGHT:
Most restaurants fall under Grade A, showing generally good hygeine compliance. Grade B and C indicate moderate to poor compliance, while N,P, and Z represent ungraded or pending inspection statueses.
